```python
import requests
```

```python
url="https://www.flipkart.com/search?q=machine+learning+books&sid=bks&as=on&as-show=on&otracker=AS_QueryStore_OrganicAutoSuggest_2_21_na_na_ps&otracker1=AS_QueryStore_OrganicAutoSuggest_2_21_na_na_ps&as-pos=2&as-type=HISTORY&suggestionId=machine+learning+books%7CBooks&requestId=83266c73-24f7-4251-a546-f65e7e8ac0f4&as-backfill=on"
```

```python
headers = {
    "User-agent": "Mozilla/5.0"
}

response = requests.get(url,headers = headers)
```

```python
print(response.status_code)
```

```python
print(response.url)
```

```python
##convert HTML into Beautifulsoup
from bs4 import BeautifulSoup
soup = BeautifulSoup(response.text, "html.parser")

print(soup.title.text)
```

```python
book_names = soup.find_all("a", class_="pIpigb")
book_prices =soup.find_all("div",class_="hZ3P6w")
```

```python
data = []
```

```python
for name, price in zip(book_names, book_prices):

    data.append({
        "Book": name.text.strip(),
        "Price": price.text.strip()
    })
```

```python
data
```

```python
import pandas as pd
df=pd.DataFrame(data)
```
